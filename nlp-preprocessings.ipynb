# Часть речи русского слова

```python
import pymorphy3

morph = pymorphy3.MorphAnalyzer()
words = ['мама', 'мыла', 'раму', 'мылом']

pos_tags = []
for word in words:
    p = morph.parse(word)[0]
    pos_tags.append((word, p.tag.POS))

print(pos_tags)
# Результат: [('Мама', 'NOUN'), ('мыла', 'VERB'), ('раму', 'NOUN'), ('мылом', 'NOUN')]
```
